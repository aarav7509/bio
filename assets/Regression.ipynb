```python
import pandas as pd

# Read the CSV file
df = pd.read_csv('train.csv')

# Display basic info about the dataset
print(df.head(5))        # First 5 rows
print(df.shape)          # (rows, columns)
```

```python
#print(df.info())          # (rows, columns)
# Count columns by data type
dtype_counts = df.dtypes.value_counts()
print(dtype_counts)
```

```python
# Select all numeric columns
numeric_df = df.select_dtypes(include=['int64', 'float64'])

#print(numeric_df.head())
#print(f"Number of numeric columns: {numeric_df.shape[1]}")
#print(numeric_df.columns.tolist())

# Full correlation matrix (all numeric variables vs each other)
corr_matrix = numeric_df.corr()

#print(corr_matrix)

saleprice_corr = corr_matrix['SalePrice'].sort_values(ascending=False)
print(saleprice_corr)
```

```python
import pandas as pd
from sklearn.feature_selection import mutual_info_regression

# Assuming 'df' is your housing dataset and 'SalePrice' is the target variable
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

# 1. Check for linear relationships (e.g., square footage)
pearson_corr = df[numeric_cols].corr(method='pearson')['SalePrice'].sort_values(ascending=False)

# 2. Check for ranked/non-linear relationships (e.g., Quality, Bedrooms)
spearman_corr = df[numeric_cols].corr(method='spearman')['SalePrice'].sort_values(ascending=False)

print("Top Pearson Linearity:\n", pearson_corr.head(5))
print("\nTop Spearman Rankings:\n", spearman_corr.head(5))
```
